```python
import pandas as pd 
```

```python
df = pd.DataFrame({
    "Name": ["Alice", "Bob", "Charlie"],
    "Age": [25, 30, 35],
    "City": ["New York", "London", "Paris"]
})
```

```python
# Print specific columns
print(df['Name'])
```

```python
# Print multiple columns
print(df[['Name', 'Age']])
```

```python

# Reading rows
print(df.iloc[0])
```

```python
#Reading the rows

# By label
print(df.loc[1, 'Name'])

# By index
print(df.iloc[1, 0])
print(df.iloc[2, 1])
print(df.iloc[0:2])
```

```python
print(df.head(2))
print(df.tail(2))
```

```python
print(df.columns.tolist())
print(df.index)
```
